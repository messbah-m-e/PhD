عملیات عنصر به عنصر

```python
A = np.array([1, 2, 3])
B = np.array([10, 20, 30])

Sum_AB = A + B
Product_AB = A * B
print(f"جمع عنصر به عنصر A + B: {Sum_AB}")     # [11 22 33]
print(f"ضرب عنصر به عنصر A * B: {Product_AB}") # [10 40 90]


Scaled_A = A * 2
print(f"ضرب A در 2: {Scaled_A}") # [2 4 6]
```

توابع آماری


ضرب ماتریسی (Dot Product)



```python
X = np.array([[1, 2], [3, 4]])
Y = np.array([[5, 6], [7, 8]])

Z = X @ Y 
print(f"ضرب ماتریسی (X @ Y):\n{Z}") 
```

حل سیستم معادلات خطی

یکی از قدرتمندترین کاربردهای جبر خطی، حل سیستم‌های معادلات خطی است (مثلاً $Ax=b$)


فرض کنید سیستم زیر را داریم:

<div dir="ltr">

$2x + 3y = 8$

$x + y = 3$
</div>

این را می‌توان به صورت ماتریسی نوشت:

<div dir="ltr">

$A = [[2, 3], [1, 1]]$

$b = [8, 3]$
</div>

 حل سیستم معادلات خطی (x = A⁻¹b)
 
</div>

```python
A = np.array([[2, 3], [1, 1]])
b = np.array([8, 3])

#
solution = np.linalg.solve(A, b)

print(f"ماتریس ضرایب A:\n{A}")
print(f"بردار جواب b: {b}")
print(f"حل (x, y): {solution}")
```

Boolean Indexing and Filtering

```python

data = np.array([10, 15, 22, 5, 8, 30, 12])

condition = data > 10 
print(f"آرایه بولین: {condition}")

filtered_data = data[condition]
print(f"داده‌های فیلتر شده: {filtered_data}")
```

Determinant

```python
B = np.array([[1, 2, 3], 
              [4, 5, 6]])

B_T = B.T
print(f"ماتریس B (2x3):\n{B}")
print(f"\nترانهاده B^T (3x2):\n{B_T}")
```

Normalization

```python
data = np.array([10, 15, 20, 25, 30])

normalized_data = (data - np.mean(data)) / np.std(data)
print(f"داده نرمال شده: {normalized_data}")
```
